```python
data={"Experience":[1,2,3,4,5],"Skills":[3,4,5,6,7],"Salary":[20000,30000,40000,50000,60000]}
df=pd.DataFrame(data)
```

```python
import numpy as np
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([20000,30000,40000,50000,70000])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y)
```

```python
poly=PolynomialFeatures()
```

```python
model1=poly.fit_transform(x)
```

```python
model=LinearRegression()
```

```python
model.fit(model1,y)
```

```python
prediction=model.predict(poly.transform([[6]]))
```

```python
print(prediction)
```
